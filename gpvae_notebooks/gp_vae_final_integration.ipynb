```python
import os

dataset_folder = 'datasets/physionet'
file_name = 'x_miss_nan.txt'
text_file = True
config_file = "imputegap/wrapper/AlgoPython/GPVAE/config_gpvae_physionet.yaml"

# dataset_folder = 'datasets/hmnist'
# dataset_folder = '/content/drive/MyDrive/gpvae/hmnist'
# file_name = 'x_miss_nan.npy'
# text_file = False
# config_file = "imputegap/wrapper/AlgoPython/GPVAE/config_gpvae_hmnist.yaml"

# dataset_folder = 'datasets/sprites'
# dataset_folder = '/content/drive/MyDrive/gpvae/sprites'
# file_name = 'x_miss_nan.npy'
# text_file = False
# config_file = "imputegap/wrapper/AlgoPython/GPVAE/config_gpvae_sprites.yaml"

if not os.path.exists(dataset_folder): os.makedirs(dataset_folder)
dataset_path = os.path.join(dataset_folder, file_name)
```

Training

```python
from imputegap.recovery.imputation import Imputation
from imputegap.recovery.manager import TimeSeries
from imputegap.tools import utils
import numpy as np

# initialize the time series object
ts = TimeSeries()

# load dataset
if text_file:
  ts.load_series(dataset_path)
else:
  npy_file = np.load(dataset_path)
  ts.import_matrix(npy_file)

params = {'config_yaml_path': config_file,
          'epoch': 2}

# impute the contaminated series
imputer = Imputation.DeepLearning.GPVAE(ts.data)
imputer.impute(params)
```

Evaluation (Physionet)

```python
from imputegap.recovery.imputation import Imputation
from imputegap.recovery.manager import TimeSeries
from imputegap.tools import utils
import numpy as np

seq_length = 48
nbr_features = 35
validation_split = 3997
validation_idx = validation_split*seq_length

# initialize the time series object
ts = TimeSeries()
ts_gt = TimeSeries()

# load dataset
if text_file:
  ts.load_series(dataset_path)
else:
  npy_file = np.load(dataset_path)
  ts.import_matrix(npy_file)

ts_gt.load_series("datasets/physionet/x_full.txt")

ground_truth = ts_gt.data[:, :validation_idx].T.reshape(-1,seq_length,nbr_features)

y_label = np.load('datasets/physionet/y_label.npy')[:validation_split]

params = {'config_yaml_path': config_file,
          'folder_checkpoint_path': 'imputegap_assets/models/20260117_130214',
          "ground_truth":ground_truth,
          "return_no_gt_imputation":True,
          "y_val":y_label,
          "verbose":False}

# impute the contaminated series
imputer = Imputation.DeepLearning.GPVAE(ts.data[:, :validation_idx])
imputer.impute(params)

imputer.recov_data.shape
print(imputer.recov_no_gt)
print(imputer.evaluation_metrics)
```
